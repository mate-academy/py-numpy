````python
python_array = [0] * 1000
python_array_size = sys.getsizeof(python_array) + sum(
    sys.getsizeof(item) for item in python_array
)
print(f"Python array size: {python_array_size} bytes")
````

````python
# Python code
def unknown_signature(matrix1, matrix2):
    result = []
    for i in range(len(matrix1)):
        row = []
        for j in range(len(matrix1[i])):
            row.append(matrix1[i][j] * matrix2[i][j])
        result.append(row)
    return result

matrix1 = [[1, 2], [3, 4]]
matrix2 = [[5, 6], [7, 8]]
%timeit unknown_signature(matrix1, matrix2)
unknown_signature(matrix1, matrix2)
````

### Task 7.2

````python
# Python code
def unknown_signature(lst1, lst2):
    bitwise_complement = [~x for x in lst1]
    bitwise_and = [x & y for x, y in zip(lst1, lst2)]
    bitwise_or = [x | y for x, y in zip(lst1, lst2)]
    return bitwise_complement, bitwise_and, bitwise_or

list1 = [5, 2, 7, 4, 9]
list2 = [3, 6, 1, 8, 10]
result_bitwise_complement, result_bitwise_and, result_bitwise_or = unknown_signature(list1, list2)
result_bitwise_complement, result_bitwise_and, result_bitwise_or
````

````python
numpy_list1 = np.array(list1)
numpy_list2 = np.array(list2)
result_bitwise_complement = np.bitwise_not(numpy_list1)
result_bitwise_and = np.bitwise_and(numpy_list1, numpy_list2)
result_bitwise_or = np.bitwise_or(numpy_list1, numpy_list2)
result_bitwise_complement, result_bitwise_and, result_bitwise_or
````

### Task 7.3 Optional

````python
# Python code
def unknown_signature(lst):
    result = []
    for num in lst:
        if num % 2 == 0 and (num % 3 == 0 or num > 10):
            modified_num = (num ^ 7) + 2
            result.append(modified_num)
    return result

my_list = [6, 9, 12, 14, 17, 20]
result_python_style = unknown_signature(my_list)
result_python_style
````

````python
numpy_values = np.array(my_list)
numpy_mask = (numpy_values % 2 == 0) & (
    (numpy_values % 3 == 0) | (numpy_values > 10)
)
result_numpy_style = np.bitwise_xor(numpy_values[numpy_mask], 7) + 2
result_numpy_style
````
