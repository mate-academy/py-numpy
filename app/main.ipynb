````python
# Python code
def unknown_signature(lst, mask, new_value):
    for i in range(len(lst)):
        if mask[i]:
            lst[i] = new_value

lst = [1, 2, 3, 4, 5]
mask = [True, False, True, False, True]
new_value = 0
unknown_signature(lst, mask, new_value)
lst
````

````python
def unknown_signature_np(arr, mask, new_value):
    arr[mask] = new_value

lst_np = np.array([1, 2, 3, 4, 5])
mask_np = np.array([True, False, True, False, True])
new_value = 0

unknown_signature_np(lst_np, mask_np, new_value)
print(lst_np)
````

### Task 7.2

````python
# Python code
def unknown_signature(lst1, lst2):
    bitwise_complement = [~x for x in lst1]
    bitwise_and = [x & y for x, y in zip(lst1, lst2)]
    bitwise_or = [x | y for x, y in zip(lst1, lst2)]
    return bitwise_complement, bitwise_and, bitwise_or

list1 = [5, 2, 7, 4, 9]
list2 = [3, 6, 1, 8, 10]
result_bitwise_complement, result_bitwise_and, result_bitwise_or = unknown_signature(list1, list2)
result_bitwise_complement, result_bitwise_and, result_bitwise_or
````

````python
def unknown_signature_np(arr1, arr2):
    bitwise_complement = np.invert(arr1)
    bitwise_and = np.bitwise_and(arr1, arr2)
    bitwise_or = np.bitwise_or(arr1, arr2)
    return bitwise_complement, bitwise_and, bitwise_or

arr1 = np.array([5, 2, 7, 4, 9])
arr2 = np.array([3, 6, 1, 8, 10])

res_not, res_and, res_or = unknown_signature_np(arr1, arr2)
print(res_not)
print(res_and)
print(res_or)
````

### Task 7.3 Optional

````python
# Python code
def unknown_signature(lst):
    result = []
    for num in lst:
        if num % 2 == 0 and (num % 3 == 0 or num > 10):
            modified_num = (num ^ 7) + 2
            result.append(modified_num)
    return result

my_list = [6, 9, 12, 14, 17, 20]
result_python_style = unknown_signature(my_list)
result_python_style
````

````python
def unknown_signature_np(arr):
    mask = (arr % 2 == 0) & ((arr % 3 == 0) | (arr > 10))
    return (arr[mask] ^ 7) + 2

my_array = np.array([6, 9, 12, 14, 17, 20])
result_numpy_style = unknown_signature_np(my_array)
print(result_numpy_style)
````
