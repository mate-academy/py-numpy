````python
# Python code
def unknown_signature(vector1, vector2):
    result = 0
    for i in range(len(vector1)):
        result += vector1[i] * vector2[i]
    return result

vector1 = [1, 2, 3]
vector2 = [4, 5, 6]

%time unknown_signature(vector1, vector2)
unknown_signature(vector1, vector2)
````

### Task 7.2

````python
# Python code
def unknown_signature(lst1, lst2):
    bitwise_complement = [~x for x in lst1]
    bitwise_and = [x & y for x, y in zip(lst1, lst2)]
    bitwise_or = [x | y for x, y in zip(lst1, lst2)]
    return bitwise_complement, bitwise_and, bitwise_or

list1 = [5, 2, 7, 4, 9]
list2 = [3, 6, 1, 8, 10]

%time unknown_signature(list1, list2)

result_bitwise_complement, result_bitwise_and, result_bitwise_or = unknown_signature(list1, list2)
result_bitwise_complement, result_bitwise_and, result_bitwise_or
````

````python
# Numpy code
def numpy_signature(lst1, lst2):
    bitwise_complement = ~lst1
    bitwise_and = lst1 & lst2
    bitwise_or = lst1 | lst2
    return bitwise_complement, bitwise_and, bitwise_or

list1 = np.array([5, 2, 7, 4, 9])
list2 = np.array([3, 6, 1, 8, 10])

%time numpy_signature(list1, list2)

result_bitwise_complement, result_bitwise_and, result_bitwise_or = numpy_signature(list1, list2)
result_bitwise_complement, result_bitwise_and, result_bitwise_or
````

### Task 7.3 Optional

````python
# Python code
def unknown_signature(lst):
    result = []
    for num in lst:
        if num % 2 == 0 and (num % 3 == 0 or num > 10):
            modified_num = (num ^ 7) + 2
            result.append(modified_num)
    return result

my_list = [6, 9, 12, 14, 17, 20]

%time unknown_signature(my_list)

result_python_style = unknown_signature(my_list)
result_python_style
````

````python
# Numpy code
def numpy_signature(lst):
    mask = (lst % 2 == 0) & ((lst % 3 == 0) | (lst > 10))

    return (lst[mask] ^ 7) + 2

my_list = np.array([6, 9, 12, 14, 17, 20])

%time numpy_signature(my_list)

result_numpy_style = numpy_signature(my_list)
result_numpy_style
````
