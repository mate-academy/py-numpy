````python
import numpy as np
import time

# Original Python Implementation
def unknown_signature(lst):
    result = []
    for num in lst:
        if num % 2 == 0 and (num % 3 == 0 or num > 10):
            modified_num = (num ^ 7) + 2
            result.append(modified_num)
    return result

# 1. Verification with small list
my_list = [6, 9, 12, 14, 17, 20]
arr = np.array(my_list)

# Generate the boolean filtering mask
mask = (arr % 2 == 0) & ((arr % 3 == 0) | (arr > 10))

# Select filtered elements and apply vectorized operation
result_np = (arr[mask] ^ 7) + 2

print("Python Result:", unknown_signature(my_list))
print("NumPy Result: ", result_np.tolist())


# 2. Performance Benchmark (Using a larger dataset)
size = 2_000_000
large_list = list(range(size))
large_arr = np.arange(size)

# --- Benchmark Python loop ---
start_py = time.perf_counter()
_ = unknown_signature(large_list)
end_py = time.perf_counter()
py_duration = end_py - start_py

# --- Benchmark NumPy vectorization ---
start_np = time.perf_counter()
large_mask = (large_arr % 2 == 0) & ((large_arr % 3 == 0) | (large_arr > 10))
_ = (large_arr[large_mask] ^ 7) + 2
end_np = time.perf_counter()
np_duration = end_np - start_np

print(f"\n--- Benchmark Results ({size:,} elements) ---")
print(f"Python loop time:  {py_duration:.6f} seconds")
print(f"NumPy vector time: {np_duration:.6f} seconds")
print(f"Vectorization speedup: {py_duration / np_duration:.1f}x faster")
````
