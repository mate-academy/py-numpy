````python
num = 1
print(sys.getsizeof(num))
````

````python
arr = [0] * 1000

print("Python list (container size):", sys.getsizeof(arr))

total_size = sys.getsizeof(arr) + sum(sys.getsizeof(x) for x in arr)
print("Python list (total size with elements):", total_size)
````

Список у Python не зберігає самі значення, а зберігає посилання на об'єкти.
У списку можуть бути елементи різних типів, тому кожен елемент є окремим об'єктом у пам’яті.

У нашому випадку список містить числа, але фактично він зберігає посилання на об'єкти типу int.

Одне посилання займає 8 байт (на 64-bit системі), тому для 1000 елементів: 1000 * 8 = 8000 байт

Додаючи службову інформацію списку (поля об'єкта Python), отримуємо ≈ 8056 байт

Значення 36056 враховує не тільки посилання, але й самі об'єкти чисел.
Розмір одного об'єкта int -> sys.getsizeof(0) = 28 байт (див. на комірку вище, де є приклад з числом 1)

Тоді -> 1000 * 28 = 28000 байт

Додаємо розмір списку -> 8056 + 28000 = 36056 байт

У результаті, список займає значно більше пам’яті, ніж здається на перший погляд.

````python
np_arr = np.array(arr, dtype=np.int8)

print("NumPy array (object size):", sys.getsizeof(np_arr))
print("NumPy array (data buffer size):", np_arr.nbytes)
````

1112 = 1000 (дані (див. нижче пояснення)) + 112 (службова інформація)

NumPy зберігає 1000 елементів послідовно (поспіль). <br>
Кожен елемент має тип int8 = 1 байт на число, тому 1000 элементів -> 1000 байт

Масив у NumPy - це не тільки дані, а об'єкт. Він зберігає багато чого, наприклад, розмірність (shape), тип даних (dtype), покажчик на дані, службову Python інформацію.

Тому виведення nbytes показує розмірність ТІЛЬКИ даних як вони є, а getsizeof - це дані + "обгортка"

Ці приклади й їхні результати пояснюють, чому Python list менш ефективний по пам’яті порівняно з NumPy, який зберігає дані компактно в одному блоці

## Task 6

Rewrite Python code to `numpy` code. Also, estimate the time of execution for both implementations.

> Please, note: `unknown_signature` this name just hide the real name of the function, to make the task more complicated.

### Task 6.1

````python
%time arr = [i for i in range(1_000_000)]
print("First five elements:", arr[:5])
print("Array shape:", len(arr))
````

````python
%time arr = np.arange(1_000_000)
print("First five elements:", arr[:5])
print("Array shape:", arr.shape)
````

Тут можна побачити перевагу NumPy масивів у порівнянні зі звичайними Python списками.

На невеликих обсягах даних (приклади нижче) різниця в продуктивності може бути незначною або навіть на користь Python списків через додаткові накладні витрати NumPy (створення масиву та виклик оптимізованих функцій).

Однак при збільшенні розміру даних (як, наприклад, у завданні 6.1) NumPy працює значно швидше, оскільки операції виконуються над масивами векторизовано та оптимізовані на рівні низькорівневих реалізацій (C).

### Task 6.2

````python
def unknown_signature(vector1, vector2):
    result = 0
    for i in range(len(vector1)):
        result += vector1[i] * vector2[i]
    return result

vector1 = [1, 2, 3]
vector2 = [4, 5, 6]
result = 0

%time result = unknown_signature(vector1, vector2)
result
````

````python
np_vector1 = np.array(vector1, dtype=np.int8)
np_vector2 = np.array(vector2, dtype=np.int8)

%time result = np_vector1 @ np_vector2
result
````

### Task 6.3

````python
def unknown_signature(matrix1, matrix2):
    result = []
    for i in range(len(matrix1)):
        row = []
        for j in range(len(matrix1[i])):
            row.append(matrix1[i][j] * matrix2[i][j])
        result.append(row)
    return result

matrix1 = [[1, 2], [3, 4]]
matrix2 = [[5, 6], [7, 8]]

%time unknown_signature(matrix1, matrix2)
````

````python
np_matrix1 = np.array(matrix1, dtype=np.int8)
np_matrix2 = np.array(matrix2, dtype=np.int8)

%time result = np_matrix1 * np_matrix2
result
````

### Task 6.4

````python
def unknown_signature(lst):
    total = 0
    for num in lst:
        total += num
    result = total / len(lst)
    return result

lst = [1, 2, 3, 4, 5]

%time result = unknown_signature(lst)
result
````

````python
np_lst = np.array(lst, dtype=np.int8)

%time result = np_lst.mean()
result
````

Як вже було зазначено, у даному випадку (завдання 6.2 - 6.4) Python показує кращий час виконання, оскільки обчислення виконуються над дуже малими обсягами даних.

Для таких простих операцій накладні витрати NumPy (створення масивів та виклик оптимізованих функцій) перевищують виграш у швидкості.

Однак при роботі з великими обсягами даних NumPy буде значно ефективнішим завдяки векторизації та використанню оптимізованих реалізацій на рівні C

Rewrite Python code to `numpy` code (Masks).

### Task 7.1

````python
def unknown_signature(lst, mask, new_value):
    for i in range(len(lst)):
        if mask[i]:
            lst[i] = new_value

lst = [1, 2, 3, 4, 5]
mask = [True, False, True, False, True]
new_value = 0
unknown_signature(lst, mask, new_value)
lst
````

### Task 7.2

````python
def unknown_signature(lst1, lst2):
    bitwise_complement = [~x for x in lst1]
    bitwise_and = [x & y for x, y in zip(lst1, lst2)]
    bitwise_or = [x | y for x, y in zip(lst1, lst2)]
    return bitwise_complement, bitwise_and, bitwise_or

list1 = [5, 2, 7, 4, 9]
list2 = [3, 6, 1, 8, 10]
result_bitwise_complement, result_bitwise_and, result_bitwise_or = unknown_signature(list1, list2)
result_bitwise_complement, result_bitwise_and, result_bitwise_or
````

````python
np_list1 = np.array(list1, dtype=np.int8)
np_list2 = np.array(list2, dtype=np.int8)

result = ~np_list1, np_list1 & np_list2, np_list1 | np_list2
result
````

### Task 7.3 Optional

````python
def unknown_signature(lst):
    result = []
    for num in lst:
        if num % 2 == 0 and (num % 3 == 0 or num > 10):
            modified_num = (num ^ 7) + 2
            result.append(modified_num)
    return result

my_list = [6, 9, 12, 14, 17, 20]
result_python_style = unknown_signature(my_list)
result_python_style
````

````python
np_my_list = np.array(my_list, dtype=np.int8)
boolean_mask = (np_my_list % 2 == 0) & ((np_my_list % 3 == 0) | (np_my_list > 10))
result_np_style = (np_my_list[boolean_mask] ^ 7) + 2
result_np_style
````
