```python
import sys
import numpy as np
```

```python
def create_null_vector(n: int) -> np.ndarray:
    """Вектор із n нулів (dtype=float за замовчуванням)."""
    return np.zeros(n, dtype=float)

def create_vector_from_range(start: int, end: int, step: int = 1) -> np.ndarray:
    """Вектор з діапазону [start, end) зі кроком step."""
    return np.arange(start, end, step, dtype=int)

def get_non_zero_indices(arr: np.ndarray) -> np.ndarray:
    """Повертає індекси ненульових елементів (одновимірний результат)."""
    return np.nonzero(arr)[0] if arr.ndim == 1 else np.transpose(np.nonzero(arr))

def create_identity_matrix(n: int) -> np.ndarray:
    """Одинична матриця n×n."""
    return np.eye(n, dtype=float)

def reverse_vector(arr: np.ndarray) -> np.ndarray:
    """Реверс елементів уздовж першої осі (1D вектор)."""
    return arr[::-1]

def create_matrix_of_ones(rows: int, cols: int, dtype=float) -> np.ndarray:
    """Матриця з одиниць розміру rows×cols."""
    return np.ones((rows, cols), dtype=dtype)

def normalize_vector(arr: np.ndarray) -> np.ndarray:
    """
    Мін–макс нормалізація 1D вектора до [0, 1].
    Якщо всі елементи однакові — повертає нулі.
    """
    arr = np.asarray(arr, dtype=float)
    mn, mx = arr.min(), arr.max()
    denom = mx - mn
    return np.zeros_like(arr) if denom == 0 else (arr - mn) / denom

def dot_product(a: np.ndarray, b: np.ndarray) -> float:
    """Скалярний добуток двох векторів однакової довжини."""
    a = np.asarray(a, dtype=float).ravel()
    b = np.asarray(b, dtype=float).ravel()
    if a.shape != b.shape:
        raise ValueError("Vectors must have the same shape")
    return float(np.dot(a, b))

def matrix_multiplication(A: np.ndarray, B: np.ndarray) -> np.ndarray:
    """Матричний добуток A @ B (перевіряє сумісність розмірів)."""
    A = np.asarray(A, dtype=float)
    B = np.asarray(B, dtype=float)
    if A.shape[-1] != B.shape[0]:
        raise ValueError("Inner dimensions must match for matrix multiplication")
    return A @ B

def unique_sorted(arr: np.ndarray) -> np.ndarray:
    """Відсортовані унікальні значення масиву."""
    return np.unique(arr)
```
