### Task 7.2

````python
# Python code
def unknown_signature(lst1, lst2):
    bitwise_complement = [~x for x in lst1]
    bitwise_and = [x & y for x, y in zip(lst1, lst2)]
    bitwise_or = [x | y for x, y in zip(lst1, lst2)]
    return bitwise_complement, bitwise_and, bitwise_or

list1 = [5, 2, 7, 4, 9]
list2 = [3, 6, 1, 8, 10]
result_bitwise_complement, result_bitwise_and, result_bitwise_or = %time unknown_signature(list1, list2)
result_bitwise_complement, result_bitwise_and, result_bitwise_or
````

````python
def np_unknown_signature(np_array1, np_array2):
    bitwise_complement = ~np_array1
    bitwise_and = np_array1 & np_array2
    bitwise_or = np_array1 | np_array2
    return bitwise_complement, bitwise_and, bitwise_or

np_array1 = np.array([5, 2, 7, 4, 9])
np_array2 = np.array([3, 6, 1, 8, 10])
result_bitwise_complement, result_bitwise_and, result_bitwise_or = %time np_unknown_signature(np_array1, np_array2)
result_bitwise_complement, result_bitwise_and, result_bitwise_or
````

### Task 7.3 Optional

````python
# Python code
def unknown_signature(lst):
    result = []
    for num in lst:
        if num % 2 == 0 and (num % 3 == 0 or num > 10):
            modified_num = (num ^ 7) + 2
            result.append(modified_num)
    return result

my_list = [6, 9, 12, 14, 17, 20]
result_python_style = %time unknown_signature(my_list)
result_python_style
````

````python
def np_unknown_signature(np_array):
    mask = (np_array % 2 == 0) & ((np_array % 3 == 0) | (np_array > 10))
    return (np_array[mask] ^ 7) + 2

np_array = np.array([6, 9, 12, 14, 17, 20])
result_python_style = %time np_unknown_signature(np_array)
result_python_style
````
