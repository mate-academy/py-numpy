````python
# Python code
def unknown_signature(vector1, vector2):
    result = 0
    for i in range(len(vector1)):
        result += vector1[i] * vector2[i]
    return result

vector1 = [1, 2, 3]
vector2 = [4, 5, 6]

%time r = unknown_signature(vector1, vector2)
````

### Task 6.3

````python
# Python code
def unknown_signature(matrix1, matrix2):
    result = []
    for i in range(len(matrix1)):
        row = []
        for j in range(len(matrix1[i])):
            row.append(matrix1[i][j] * matrix2[i][j])
        result.append(row)
    return result

matrix1 = [[1, 2], [3, 4]]
matrix2 = [[5, 6], [7, 8]]

%time r = unknown_signature(matrix1, matrix2)
````

````python
# numpy code
np_m1 = np.array(matrix1)
np_m2 = np.array(matrix2)

%time r = np_m1 * np_m2
````

## Task 7

Rewrite Python code to `numpy` code (Masks).

### Task 7.1

````python
# Python code
def unknown_signature(lst, mask, new_value):
    for i in range(len(lst)):
        if mask[i]:
            lst[i] = new_value

lst = [1, 2, 3, 4, 5]
mask = [True, False, True, False, True]
new_value = 0
unknown_signature(lst, mask, new_value)
lst
````

````python
# numpy code
np_lst = np.array([1, 2, 3, 4, 5])
np_mask = np.array([True, False, True, False, True])
np_lst[np_mask] = 0
np_lst
````

### Task 7.2

````python
# Python code
def unknown_signature(lst1, lst2):
    bitwise_complement = [~x for x in lst1]
    bitwise_and = [x & y for x, y in zip(lst1, lst2)]
    bitwise_or = [x | y for x, y in zip(lst1, lst2)]
    return bitwise_complement, bitwise_and, bitwise_or

list1 = [5, 2, 7, 4, 9]
list2 = [3, 6, 1, 8, 10]
result_bitwise_complement, result_bitwise_and, result_bitwise_or = unknown_signature(list1, list2)
result_bitwise_complement, result_bitwise_and, result_bitwise_or
````

````python
# numpy code
arr1 = np.array(list1)
arr2 = np.array(list2)
np_complement = ~arr1
np_and = arr1 & arr2
np_or = arr1 | arr2
np_complement, np_and, np_or
````

### Task 7.3 Optional

````python
# Python code
def unknown_signature(lst):
    result = []
    for num in lst:
        if num % 2 == 0 and (num % 3 == 0 or num > 10):
            modified_num = (num ^ 7) + 2
            result.append(modified_num)
    return result

my_list = [6, 9, 12, 14, 17, 20]
result_python_style = unknown_signature(my_list)
result_python_style
````

````python
# numpy code
np_list = np.array(my_list)
mask = (np_list % 2 == 0) & ((np_list % 3 == 0) | (np_list > 10))
result_numpy_style = (np_list[mask] ^ 7) + 2
result_numpy_style
````
